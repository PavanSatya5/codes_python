# **Student Score Predictor Using ML**

```python
import numpy as n
import matplotlib.pyplot as p
hours=n.array([1,2,3,4,5])
score=n.array([30,40,50,65,80])
p.scatter(hours,score)
p.show()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression # this how we import the linear regression model in sklearn
from sklearn.model_selection import train_test_split# this is used to split the data for training and testing
from sklearn.metrics import mean_squared_error,root_mean_squared_error,mean_absolute_error,r2_score
input=hours.reshape(-1,1)
'''
      assigning the input labels,
      The inputs always should be in 2- array format for classical ML,
      and also here our dataset contains 5students and 1 output column for that we'll do the reshape here
      for reshape(-1,1) into -1 means:
      NumPy, YOU figure out the correct number of rows
      if u dont do the reshape here then the model will u assume that it is one student having 5features....
'''
output=score#assigning the output labels
X_train,X_test,Y_train,Y_test=train_test_split(input,output,test_size=0.2,random_state=42)
'''test_size is the testing on the 20% of the data and random state=42 random_state=42 Does

It fixes the randomness.

So every time you run the code:

Same split

Same training data

Same test data

Same result
'''
model=LinearRegression()#creating the model
model.fit(X_train,Y_train)#this is the training part
#now lets see about the prediction of this...
#print(model.n_features_in_) ---> this is used to determine the no-of-values that ml model expects during the prediction
predOutput=model.predict([[6]])#giving the last value, which we have left out of five
print(f'this is the predicted value:{predOutput} ')
predY=model.predict(X_test)#predicted value for evaluating metrics
mse=mean_squared_error(Y_test,predY)#here the Y_test is the actual value and predY is the predicted value
print(f'MSE:{mse}')
print(f'RMSE:{root_mean_squared_error(Y_test,predY)} and {n.sqrt(mse)}')
print(f'MAE:{mean_absolute_error(Y_test,predY)}')
print(f'R2Score:{r2_score(Y_test,predY)}')
```
